```python
"""
Leak-free rewrite of 06_ml_signal_classifier.ipynb  (UNTESTED: I don't have your CSV).

Changes vs the original:
1. Target is the *forward* 5-day return sign, not a same-day function of MACD/RSI.
2. Rolling / pct_change / shift are computed per ticker (no bleed between stocks).
3. Chronological train/test split with a gap, instead of a random split.
4. Features are scale-free (ratios), so they are comparable across stocks.
5. SMOTE dropped: the new target is roughly balanced.
"""
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, classification_report

PATH = "../data/nifty50_features.csv"
TICKER = "Ticker"   # matches the column written by 02_feature_engineering.ipynb
H = 5               # forward horizon in trading days

df = pd.read_csv(PATH, parse_dates=["Date"])
df = df.sort_values([TICKER, "Date"]).reset_index(drop=True)
g = df.groupby(TICKER)

# --- Features (all use only information available at day t) ---
df["MACD_n"] = df["MACD"] / df["Close"]
df["MACD_Signal_n"] = df["MACD_Signal"] / df["Close"]
df["MACD_Hist_n"] = df["MACD_n"] - df["MACD_Signal_n"]
df["Close_vs_SMA20"] = df["Close"] / df["SMA_20"] - 1
df["Close_vs_EMA20"] = df["Close"] / df["EMA_20"] - 1
df["Daily_Return"] = g["Close"].pct_change()
df["Volatility_5"] = df.groupby(TICKER)["Daily_Return"].transform(lambda s: s.rolling(5).std())
df["Momentum_10"] = df["Close"] / g["Close"].shift(10) - 1
df["Volume_ratio"] = df["Volume"] / g["Volume"].transform(lambda s: s.rolling(20).mean())

features = [
    "MACD_n", "MACD_Signal_n", "MACD_Hist_n", "RSI",
    "Close_vs_SMA20", "Close_vs_EMA20",
    "Volatility_5", "Momentum_10", "Volume_ratio",
]

# --- Target: does the stock go up over the next H trading days? ---
df["Fwd_Ret"] = g["Close"].shift(-H) / df["Close"] - 1
df["Target"] = (df["Fwd_Ret"] > 0).astype(int)
df = df.dropna(subset=features + ["Fwd_Ret"])

# --- Chronological split with a gap (labels look H days ahead, so avoid overlap) ---
cutoff = df["Date"].quantile(0.8)
train = df[df["Date"] <= cutoff - pd.Timedelta(days=10)]
test = df[df["Date"] > cutoff]

model = RandomForestClassifier(
    n_estimators=300, min_samples_leaf=50, class_weight="balanced",
    random_state=42, n_jobs=-1,
)
model.fit(train[features], train["Target"])

prob = model.predict_proba(test[features])[:, 1]
print("Train rows:", len(train), "| Test rows:", len(test))
print("Test base rate (up-moves):", round(test["Target"].mean(), 3))
print("Test ROC AUC:", round(roc_auc_score(test["Target"], prob), 3))
print(classification_report(test["Target"], (prob > 0.5).astype(int)))
```
